```python
import os
import cv2
from google.colab.patches import cv2_imshow
from ultralytics import YOLO

# 1. Carrega os dois modelos
print("A carregar os modelos...")
modelo_base = YOLO("yolo11n.pt")                    # Modelo oficial para as classes gerais
modelo_personalizado = YOLO("runs/detect/train/weights/best.pt") # O teu modelo treinado

# 2. Define o caminho da imagem
caminho_imagem = "Gun_Test.jpg"

# Valida se o ficheiro existe antes de continuar
if not os.path.exists(caminho_imagem):
    print(f"\n[ERRO] O ficheiro '{caminho_imagem}' não foi encontrado. Faz o upload da imagem para o Colab.")
else:
    # 3. Lê a imagem usando o OpenCV (em formato BGR)
    imagem = cv2.imread(caminho_imagem)
    print(f"\nImagem '{caminho_imagem}' carregada com sucesso!")
    print("A executar a inferência em ambos os modelos e a desenhar as caixas para todas as classes...")

    # 4. Executa a inferência nos dois modelos
    resultados_base = modelo_base(imagem)
    resultados_personalizado = modelo_personalizado(imagem)

    # Lista unificada para guardar as deteções em texto
    todas_deteccoes = []

    # --- DESENHAR AS CAIXAS DO MODELO BASE (Todas as classes detetadas em AZUL) ---
    for r in resultados_base:
        for box in r.boxes:
            cls_id = int(box.cls[0])
            cls_name = modelo_base.names[cls_id]

            conf = float(box.conf[0])
            b = box.xyxy[0].tolist() # [x_min, y_min, x_max, y_max]
            x1, y1, x2, y2 = int(b[0]), int(b[1]), int(b[2]), int(b[3])

            # Desenha a caixa no OpenCV (Cor Azul: (255, 0, 0)) para qualquer classe detetada
            cv2.rectangle(imagem, (x1, y1), (x2, y2), (255, 0, 0), 2)
            cv2.putText(imagem, f"{cls_name} {conf:.2f}", (x1, y1 - 10),
                        cv2.FONT_HERSHEY_SIMPLEX, 0.5, (255, 0, 0), 2)

            todas_deteccoes.append({"modelo": "Base", "classe": cls_name, "confianca": conf, "box": b})

    # --- DESENHAR AS CAIXAS DO MODELO PERSONALIZADO (Ex: GUN em VERMELHO) ---
    for r in resultados_personalizado:
        for box in r.boxes:
            cls_id = int(box.cls[0])
            cls_name = modelo_personalizado.names[cls_id]
            conf = float(box.conf[0])
            b = box.xyxy[0].tolist()
            x1, y1, x2, y2 = int(b[0]), int(b[1]), int(b[2]), int(b[3])

            # Desenha a caixa no OpenCV (Cor Vermelha: (0, 0, 255))
            cv2.rectangle(imagem, (x1, y1), (x2, y2), (0, 0, 255), 2)
            cv2.putText(imagem, f"{cls_name} {conf:.2f}", (x1, y1 - 10),
                        cv2.FONT_HERSHEY_SIMPLEX, 0.5, (0, 0, 255), 2)

            todas_deteccoes.append({"modelo": "Personalizado", "classe": cls_name, "confianca": conf, "box": b})

    # 5. Exibe os resultados em texto no terminal
    print(f"\n--- Total de Objetos Detetados: {len(todas_deteccoes)} ---")
    for d in todas_deteccoes:
        print(f"[{d['modelo']}] Classe: {d['classe']} | Confiança: {d['confianca']:.2f}")

    # 6. Mostra a imagem final combinada diretamente no output do Colab
    print("\nA exibir a imagem final com todas as deteções unificadas:")
    cv2_imshow(imagem)
```

```python
%%writefile main.py
from fastapi import FastAPI, UploadFile, File, HTTPException
from ultralytics import YOLO
import cv2
import numpy as np

app = FastAPI(
    title="API de Monitoramento Residencial",
    version="1.0.0"
)

# Caminhos dos modelos no Colab
MODELO_BASE_PATH = "yolo11n.pt"
MODELO_ARMAS_PATH = "runs/detect/train/weights/best.pt"

print("A carregar os modelos...")
modelo_base = YOLO(MODELO_BASE_PATH)
modelo_armas = YOLO(MODELO_ARMAS_PATH)
print("Modelos carregados com sucesso!")

CLASSES_RUA = ['person', 'car', 'motorcycle', 'bicycle', 'bus', 'truck']

@app.get("/")
def home():
    return {"status": "online", "message": "API a funcionar no Colab!"}

@app.post("/detect")
async def detect_objects(file: UploadFile = File(...)):
    if not file.content_type.startswith("image/"):
        raise HTTPException(status_code=400, detail="O arquivo enviado não é uma imagem válida.")

    try:
        contents = await file.read()
        nparr = np.frombuffer(contents, np.uint8)
        image = cv2.imdecode(nparr, cv2.IMREAD_COLOR)

        if image is None:
            raise HTTPException(status_code=400, detail="Erro ao decodificar a imagem.")

        resultados_base = modelo_base(image)
        resultados_armas = modelo_armas(image)

        detections = []

        for r in resultados_base:
            for box in r.boxes:
                cls_id = int(box.cls[0])
                cls_name = modelo_base.names[cls_id]
                conf = float(box.conf[0])
                if cls_name in CLASSES_RUA:
                    b = box.xyxy[0].tolist()
                    detections.append({
                        "class_name": cls_name,
                        "confidence": round(conf, 2),
                        "box": {"x_min": int(b[0]), "y_min": int(b[1]), "x_max": int(b[2]), "y_max": int(b[3])}
                    })

        for r in resultados_armas:
            for box in r.boxes:
                cls_id = int(box.cls[0])
                cls_name = modelo_armas.names[cls_id]
                conf = float(box.conf[0])
                b = box.xyxy[0].tolist()
                detections.append({
                    "class_name": cls_name,
                    "confidence": round(conf, 2),
                    "box": {"x_min": int(b[0]), "y_min": int(b[1]), "x_max": int(b[2]), "y_max": int(b[3])}
                    })

        return {"status": "success", "detections": detections}

    except Exception as e:
        raise HTTPException(status_code=500, detail=str(e))
```
